```python
from datetime import date
from ipynb.fs.defs.predict import predict_signal
from ipynb.fs.defs.data import reinforcement_data
def portfolio_recommendations(portfolio, cash):

    recommendations = []

    portfolio_value = (cash+ sum(stock["position_value"]for stock in portfolio))

    for stock in portfolio:

        ticker = stock["ticker"]

        current_price = float(
            reinforcement_data(ticker,"2024-01-01",str(date.today())).iloc[-1])

        position_fraction = (stock["position_value"]/ portfolio_value)

        unrealized_pnl = (current_price- stock["avg_buy_price"]) / stock["avg_buy_price"]

        signal = predict_signal(
            ticker=ticker,
            position_fraction=position_fraction,
            unrealized_pnl=unrealized_pnl
        )

        recommendations.append({
            "ticker": ticker,
            "action": signal["action"],
            "price": current_price,
            "position_value": stock["position_value"],
            "portfolio_weight": round(position_fraction * 100, 2),
            "unrealized_pnl": round(unrealized_pnl * 100, 2)
        })

    return recommendations
```

```python
#example
portfolio = [
    {
        "ticker": "AAPL",
        "position_value": 3000,
        "avg_buy_price": 200
    },
    {
        "ticker": "MSFT",
        "position_value": 2000,
        "avg_buy_price": 400
    }
]

cash = 5000

portfolio_recommendations(
    portfolio,
    cash
)
```
