```python
import torch
import torch.nn as nn

class QNetwork(nn.Module):

    def __init__(self):

        super().__init__()
        self.fc1 = nn.Linear(11,64)
        self.fc2 = nn.Linear(64,64)
        self.fc3 = nn.Linear(64,5)

    def forward(self,z):
        x = self.fc1(z)
        x = torch.relu(x)
        x = self.fc2(x)
        x = torch.relu(x)
        x = self.fc3(x)

        return x
```
